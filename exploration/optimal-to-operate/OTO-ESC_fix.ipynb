# Optimal to operate: one tuned ESC on random drifting plants (100k runs)

**Question.** With one fixed, well-tuned set of ESC parameters, how often and by how much does ESC beat the
fixed ratio $r = 0.5$ when the plant is uncertain?

**Data.** `OTO-ESC_fix-1_0.2_6-100kshots.csv`, generated by `OTO_ESC_fix_Plant_rand.m`: 100,000 runs of 100 s
with the same ESC parameters in every run ($k = 1$, $a = 0.2$, $f = 6$ Hz, chosen from the 3-D plots in
`OTO-Plant_fix.ipynb`) and a random plant with $r_{\mathrm{ideal},0} \in [0.45, 0.55]$ and
$\lambda \in [-0.002, 0.002]$ s$^{-1}$. The estimate $\hat r$ starts at 0.5.

#### Added plots: share of plants on which ESC wins

The same plant map as a contour plot. The colour is the share of plants in each region of the map on which ESC gives
more energy than $r = 0.5$: the plants are put in bins (about 20 per bin), and the share per bin is smoothed over
1.5 bins, as in the other heat maps of this repository. Left: the whole range of plants. Right: zoomed on the region
where ESC loses, with finer bins (about 10 plants per bin). Dotted: plants whose ideal ratio is 0.5 on average over
the run. Section 8 explains the shape of the region.

Within one bin the outcome is not random: ESC either wins or loses on each plant. A share between 0 and 100 % means
that the boundary of the losing region runs through that bin, so the contours show where the boundary is, blurred
by the bin size.

```python
WIN_LEVELS = [0, 10, 25, 50, 75, 90, 99, 100]


def run_average_half(lam_e3, T=esc.T_END):
    """r_ideal_0 for which r_ideal(t) = r_ideal_0 exp(lambda t) is 0.5 on average over the run."""
    lam = np.asarray(lam_e3, dtype=float) * 1e-3
    growth = np.where(np.abs(lam) < 1e-12, 1.0, np.expm1(lam * T) / np.where(np.abs(lam) < 1e-12, 1.0, lam * T))
    return 0.5 / growth


def win_share_map(runs, r_range, lam_range_e3, runs_per_bin):
    """Smoothed share [%] of plants on which ESC wins, on a grid over the plant map."""
    inside = runs['r_ideal_0'].between(*r_range) & (1e3 * runs['lambda_r_ideal']).between(*lam_range_e3)
    part = runs[inside]
    n_bins = esc.bins_for(len(part), runs_per_bin=runs_per_bin)
    r_edges = np.linspace(*r_range, n_bins + 1)
    l_edges = np.linspace(*lam_range_e3, n_bins + 1)
    share = 100 * esc.binned_map(part['r_ideal_0'], 1e3 * part['lambda_r_ideal'], part['choose_ESC'], r_edges, l_edges)
    return 0.5 * (r_edges[1:] + r_edges[:-1]), 0.5 * (l_edges[1:] + l_edges[:-1]), share, n_bins


def win_contour(ax, r_range, lam_range_e3, runs_per_bin, title):
    r_mid, l_mid, share, n_bins = win_share_map(df, r_range, lam_range_e3, runs_per_bin)
    filled = ax.contourf(r_mid, l_mid, share.T, levels=WIN_LEVELS, cmap='RdYlBu')
    lines = ax.contour(r_mid, l_mid, share.T, levels=[50], colors='black', linewidths=1.0)
    ax.clabel(lines, fmt='%g %%', fontsize=8)
    lam_line = np.linspace(*lam_range_e3, 200)
    ax.plot(run_average_half(lam_line), lam_line, color=esc.COLORS['grey'], ls=':', lw=1.2)
    ax.set_xlim(r_mid[0], r_mid[-1]); ax.set_ylim(l_mid[0], l_mid[-1])   # bin centres: no empty strip at the edges
    ax.set_xlabel(esc.label('r_ideal_0'))
    ax.set_title(f'{title} ({n_bins} x {n_bins} bins)')
    return filled


FULL_RANGE = ((0.45, 0.55), (-2.0, 2.0))
ZOOM_RANGE = ((0.478, 0.512), (-0.6, 0.9))

# dpi=200: sharp when the notebook shrinks the figure
fig, axes = plt.subplots(1, 2, figsize=(14, 5), layout='constrained', dpi=200)
filled = win_contour(axes[0], *FULL_RANGE, runs_per_bin=20, title='All plants')
win_contour(axes[1], *ZOOM_RANGE, runs_per_bin=10, title='Zoom on the losing region')
axes[0].add_patch(plt.Rectangle((ZOOM_RANGE[0][0], ZOOM_RANGE[1][0]), np.diff(ZOOM_RANGE[0])[0], np.diff(ZOOM_RANGE[1])[0],
                                fill=False, edgecolor='black', lw=0.8, ls='--'))
for ax in axes:
    ax.set_ylabel(r'drift rate $\lambda$ [10$^{-3}$ s$^{-1}$]')
fig.colorbar(filled, ax=axes, label='plants on which ESC wins [%]', ticks=WIN_LEVELS)
fig.suptitle('Share of plants on which ESC beats r = 0.5 (k = 1, a = 0.2, f = 6 Hz)')
plt.show()

print(f"ESC wins on {df['choose_ESC'].mean():.1%} of all plants.")
for width in (0.01, 0.005, 0.0025):
    near = df[(df['r_ideal_0'] - run_average_half(1e3 * df['lambda_r_ideal'])).abs() < width]
    print(f"Plants whose initial ratio is within {width} of the 'run average = 0.5' line: {len(near):,}; ESC wins on {near['choose_ESC'].mean():.1%}.")
```

## 6. Chosen ESC sets on random plants

The sets below come from the fixed-plant analysis. The reference set was read off the 3-D plots in
`OTO-Plant_fix.ipynb`. The other four were picked in `OTO-Plant_fix-paired.ipynb` (Section 6) among the sets with
$\eta \ge 0.9$ on all eight fixed plants that settle within 0.01 on every plant with a late tracking error below 0.005:

| set | $k$ | $a$ | $f$ [Hz] | rule |
|---|---|---|---|---|
| reference | 1 | 0.2 | 6 | from the 3-D plots |
| fastest | 0.255 | 0.0517 | 8.41 | shortest worst settling time |
| fastest, $a \ge 0.15$ | 0.265 | 0.192 | 9.09 | the same, with a dither that would still be visible in a noisy signal |
| fastest, $f \le 3$ Hz | 0.0982 | 0.0556 | 2.97 | the same, with a slow dither |
| best energy, $a \ge 0.1$ | 1.6 | 0.11 | 9.91 | highest worst-plant energy score with $a \ge 0.1$ |

**Data.** `OTO_ESC_chosen_Plant_rand.py` runs every set on the **same** 100,000 random plants: the plants of the MATLAB
data set above, row by row. The settings are those of the paired sweep, not of the MATLAB script: every run starts
from a plasma that is already running (full density, steady state at $r = 0.5$), and the settling time and tracking
error are measured on the ratio the plasma receives. A separate file holds two references per plant: the energy
with $r = 0.5$ and the energy when the ratio follows $r_\mathrm{ideal}(t)$ exactly.

Three energies per run, all in the same units as above:

- **gain** $= E_\mathrm{ESC} - E_{r=0.5}$ (negative: ESC loses);
- **margin** $= E_\mathrm{ideal} - E_{r=0.5}$: what the fixed ratio loses against the best possible fuelling;
- **shortfall** $= E_\mathrm{ideal} - E_\mathrm{ESC}$: what ESC loses against the best possible fuelling.

So gain = margin − shortfall: ESC loses exactly on the plants where its shortfall is larger than the margin.

```python
import importlib.util
from IPython.display import display

# the list of ESC sets is read from the script that generated the data, so it is written down only once
spec = importlib.util.spec_from_file_location('chosen_sweep', 'OTO_ESC_chosen_Plant_rand.py')
chosen_sweep = importlib.util.module_from_spec(spec)
spec.loader.exec_module(chosen_sweep)

references = pd.read_csv('OTO-ESC_chosen-references-100k.csv')
assert np.allclose(references[esc.PLANT_PARAMS], df[esc.PLANT_PARAMS], rtol=0, atol=1e-15)  # the plants of the MATLAB data set

chosen = {}
for name, k, a, f in chosen_sweep.esc_sets:
    runs = pd.read_csv(f'OTO-ESC_chosen-{chosen_sweep.set_tag(k, a, f)}-100k.csv')
    assert np.array_equal(runs[esc.PLANT_PARAMS].to_numpy(), references[esc.PLANT_PARAMS].to_numpy())  # same plants, same order
    runs = runs.join(references[['energy_const', 'energy_ideal']])
    runs['gain'] = runs['energy_ESC'] - runs['energy_const']
    runs['margin'] = runs['energy_ideal'] - runs['energy_const']
    runs['shortfall'] = runs['energy_ideal'] - runs['energy_ESC']
    chosen[name] = runs
SET_COLORS = dict(zip(chosen, esc.CYCLE))

summary = pd.DataFrame({
    name: {'k': r['optimizer_gain'].iloc[0], 'a': r['dither_amplitude'].iloc[0], 'f [Hz]': r['dither_frequency'].iloc[0],
           'lam_avg [1/s]': esc.lam_avg(r['optimizer_gain'].iloc[0], r['dither_frequency'].iloc[0]),
           'ESC better [%]': 100 * (r['gain'] > 0).mean(),
           'median gain': r['gain'].median(),
           'median shortfall': r['shortfall'].median(),
           'largest shortfall': r['shortfall'].max(),
           'settled within 0.01 [%]': 100 * r['settling_time_0.01'].notna().mean(),
           'median settling time [s]': r['settling_time_0.01'].median(),
           'longest settling time [s]': r['settling_time_0.01'].max(),
           'median tracking error, 50-100 s': r['tracking_error_late'].median(),
           'largest tracking error, 50-100 s': r['tracking_error_late'].max()}
    for name, r in chosen.items()})
print(f"{len(references):,} plants; margin of the fixed ratio: median {(references['energy_ideal'] - references['energy_const']).median():.2f}, "
      f"largest {(references['energy_ideal'] - references['energy_const']).max():.2f} energy units")
summary.round(4)
```

## 7. How much does ESC lose when it loses?

**Expected size of a loss.** If $r_\mathrm{ideal}$ stays at 0.5, the fixed ratio is already optimal and ESC can only
lose. What it loses is the dither: in steady state a ratio error $\Delta$ costs $n_\mathrm{ref}^2 \Delta^2$ of the
power, the plasma smooths the dither $a \sin(\omega t)$ with its loss rate $\bar\alpha$, and over a run of $T$ = 100 s
this adds up to

$$L_\mathrm{dith} = T\,\frac{n_\mathrm{ref}^2}{4} \cdot 2a^2 \frac{\bar\alpha^2}{\bar\alpha^2 + \omega^2} \qquad (\texttt{esc.dither\_loss}).$$

Away from 0.5 the fixed ratio loses its margin, which grows like $(r_\mathrm{ideal} - 0.5)^2$. ESC should therefore
lose only on the plants whose margin is smaller than $L_\mathrm{dith}$, and then by at most about $L_\mathrm{dith}$.
For $T n_\mathrm{ref}^2 = 13{,}225$ the break-even distance is $|r_\mathrm{ideal} - 0.5| \approx \sqrt{L_\mathrm{dith} / 13{,}225}$.

```python
T_N2 = esc.T_END * esc.N_REF ** 2
loss_rows = {}
for name, r in chosen.items():
    k, a, f = (r[c].iloc[0] for c in esc.ESC_PARAMS)
    lost = r[r['gain'] < 0]
    L = float(esc.dither_loss(a, f))
    loss_rows[name] = {
        'predicted largest loss L_dith': L,
        'largest loss': -r['gain'].min(),
        'largest loss / L_dith': -r['gain'].min() / L,
        'largest loss [% of output]': -100 * (r['gain'] / r['energy_const']).min(),
        'runs where ESC loses': len(lost),
        'share [%]': 100 * len(lost) / len(r),
        'median loss': -lost['gain'].median(),
        'largest margin among them': lost['margin'].max(),
        'predicted break-even |r_ideal - 0.5|': np.sqrt(L / T_N2),
        'largest gain': r['gain'].max(),
        'largest gain / largest loss': r['gain'].max() / -r['gain'].min(),
    }
loss_table = pd.DataFrame(loss_rows)
loss_table.round(5)
```

## 8. Why the losing region is a tilted ellipse

In Section 5 the plants on which ESC loses form a tilted ellipse around $r_{\mathrm{ideal},0} = 0.5$, $\lambda = 0$, not a
circle. This section explains its shape, angle and size.

**Condition.** ESC loses where the fixed ratio loses less than ESC itself does: margin < shortfall (Section 7). Near
$r_\mathrm{ideal} = 0.5$ the shortfall $S_0$ of ESC is almost the same on every plant (the dither loss, plus a start-up
cost with the MATLAB start). The margin is what $r = 0.5$ loses against perfect fuelling. In steady state a ratio error
$\Delta$ costs $n_\mathrm{ref}^2 \Delta^2$ of the power, so

$$\text{margin} \approx T\, n_\mathrm{ref}^2 \,\big\langle (r_\mathrm{ideal}(t) - 0.5)^2 \big\rangle ,$$

averaged over the run of length $T$.

**Why an ellipse.** Write $u = r_{\mathrm{ideal},0} - 0.5$ (starting offset) and $v = r_{\mathrm{ideal},0}\,\lambda\, T$ (how
much the ideal ratio drifts during the run). With $r_\mathrm{ideal}(t) \approx r_{\mathrm{ideal},0}\,(1 + \lambda t)$:

$$\big\langle (r_\mathrm{ideal} - 0.5)^2 \big\rangle = \underbrace{(u + v/2)^2}_{\text{run average away from 0.5}}
+ \underbrace{v^2/12}_{\text{spread around the average}} = u^2 + uv + v^2/3 .$$

ESC loses where $u^2 + uv + v^2/3 < \delta_*^2$ with $\delta_* = \sqrt{S_0 / (T n_\mathrm{ref}^2)}$: inside an ellipse.

**Why it is tilted and not a circle.** The cross term $uv$: a drift can cancel a starting offset. A plant that starts
below 0.5 and drifts up spends its run around 0.5, so $r = 0.5$ is nearly right for it. Along the line $u = -v/2$
(run average exactly 0.5) only the spread $v^2/12$ counts and the ellipse is long; across that line the full offset of
the average counts and the ellipse is narrow. In $(u, v)$ the long axis has slope $-1.87$ and the axes have the ratio
4.4, whatever the ESC set, the start or the run length.

**What sets the angle in the plot.** On the plotted axes $\lambda = v / (r_{\mathrm{ideal},0}\, T)$, so the long axis has
the slope $\mathrm{d}\lambda / \mathrm{d}r_{\mathrm{ideal},0} \approx -1.87 / (0.5\,T)$: $-37 \cdot 10^{-3}$ s$^{-1}$ per unit
of $r_{\mathrm{ideal},0}$ for $T$ = 100 s. Only the **run length** sets it: a longer run gives the same drift more time
to move the ideal ratio, so the ellipse becomes flatter. (How tilted it *looks* also depends on the axis ranges of the
plot.)

**What sets the size.** $\delta_*$: the half-width in $r_{\mathrm{ideal},0}$ at $\lambda = 0$ is $\delta_*$, and the ends of
the long axis are at $u = \mp\sqrt{3}\,\delta_*$, $\lambda = \pm \sqrt{12}\,\delta_* / (r_{\mathrm{ideal},0} T)$. The ESC
parameters enter only here, through $S_0$ (for the steady start $S_0 \approx L_\mathrm{dith} \propto a^2$, Section 7).
$n_\mathrm{ref}^2$ cancels, because the margin and the dither loss both scale with it.

**What sets the position.** The ellipse is centred on $(0.5, 0)$ unless something prefers one side of 0.5. The MATLAB
start does: the plasma starts deuterium-rich, so during the start-up a slightly tritium-rich mix is better, and $r = 0.5$
loses less on plants just below 0.5 than just above it. This adds a term $\beta u$ to the margin, which moves the
centre to $u_c = -2\beta / (T n_\mathrm{ref}^2)$, $v_c = -1.5\,u_c$. Because the ellipse is long, a small $\beta$ moves it far.

```python
T_RUN = esc.T_END


def mean_exp(c, T):
    """Run average of exp(c t) over 0..T (1 for c = 0)."""
    c = np.asarray(c, dtype=float)
    small = np.abs(c) < 1e-12
    return np.where(small, 1.0, np.expm1(c * T) / np.where(small, 1.0, c * T))


def mean_square_offset(r0, lam, T=T_RUN):
    """Run average of (r_ideal(t) - 0.5)^2 for r_ideal(t) = r0 exp(lam t), exact."""
    r0, lam = np.asarray(r0, dtype=float), np.asarray(lam, dtype=float)
    return r0 ** 2 * mean_exp(2 * lam, T) - r0 * mean_exp(lam, T) + 0.25


def predicted_margin(r0, lam, T=T_RUN, beta=0.0):
    """Energy the fixed ratio r = 0.5 loses against perfect fuelling (beta: start asymmetry, see above)."""
    return T * esc.N_REF ** 2 * mean_square_offset(r0, lam, T) + beta * (np.asarray(r0) - 0.5)


def losing_axes(r0, lam, T=T_RUN):
    """Slope of the long axis and ratio of the axes of a cloud of plants, in u = r0 - 0.5, v = r0 lam T."""
    u, v = np.asarray(r0) - 0.5, np.asarray(r0) * np.asarray(lam) * T
    w, V = np.linalg.eigh(np.cov(np.vstack([u, v])))
    return V[1, 1] / V[0, 1], np.sqrt(w[1] / w[0])


Q_INV_W, Q_INV_V = np.linalg.eigh(np.linalg.inv([[1, 0.5], [0.5, 1 / 3]]))   # u^2 + uv + v^2/3
PRED_SLOPE, PRED_RATIO = Q_INV_V[1, 1] / Q_INV_V[0, 1], np.sqrt(Q_INV_W[1] / Q_INV_W[0])

# start asymmetry of the MATLAB start, measured on the grid of plants (file described below)
grid_matlab = pd.read_csv('OTO-losing_region-matlab-T100.csv')
margin_grid = grid_matlab['energy_ideal'] - grid_matlab['energy_const']
near_half = predicted_margin(grid_matlab['r_ideal_0'], grid_matlab['lambda_r_ideal']) < 5
fit, *_ = np.linalg.lstsq(np.column_stack([predicted_margin(grid_matlab['r_ideal_0'], grid_matlab['lambda_r_ideal'])[near_half],
                                           grid_matlab['r_ideal_0'][near_half] - 0.5]), margin_grid[near_half], rcond=None)
BETA_MATLAB = fit[1]
centre = np.isclose(grid_matlab['r_ideal_0'], 0.5, rtol=0, atol=1e-9) & np.isclose(grid_matlab['lambda_r_ideal'], 0, rtol=0, atol=1e-12)
S0_MATLAB = float((grid_matlab['energy_ideal'] - grid_matlab['energy_ESC'])[centre].iloc[0])

# the losing plants of Section 5 (MATLAB start) and of the chosen sets (steady start), against the prediction
cases = {'reference, MATLAB start': (df.assign(gain=df['energy_ESC'] - df['energy_const']), S0_MATLAB, BETA_MATLAB)}
for name, r in chosen.items():
    cases[f'{name}, steady start'] = (r, float(esc.dither_loss(r['dither_amplitude'].iloc[0], r['dither_frequency'].iloc[0])), 0.0)

rows = {}
for name, (r, S0, beta) in cases.items():
    lost = r[r['gain'] < 0]
    predicted = predicted_margin(r['r_ideal_0'], r['lambda_r_ideal'], beta=beta) < S0
    slope, ratio = losing_axes(lost['r_ideal_0'], lost['lambda_r_ideal'])
    rows[name] = {'S0': S0, 'delta_*': np.sqrt(S0 / (T_RUN * esc.N_REF ** 2)), 'plants where ESC loses': len(lost),
                  'predicted': int(predicted.sum()), 'prediction right [%]': 100 * (predicted == (r['gain'] < 0)).mean(),
                  'long-axis slope in (u, v)': slope, 'axis ratio': ratio,
                  'centre r_ideal_0': lost['r_ideal_0'].mean(), 'centre lambda [1e-3/s]': 1e3 * lost['lambda_r_ideal'].mean()}
ellipse_table = pd.DataFrame(rows).T
print(f'Prediction for every case: long-axis slope {PRED_SLOPE:.2f}, axis ratio {PRED_RATIO:.2f}.')
print(f'MATLAB start: S0 = {S0_MATLAB:.3f} (dither loss {float(esc.dither_loss(1 * 0.2, 6)):.3f}), beta = {BETA_MATLAB:.1f}; '
      f'predicted centre r_ideal_0 = {0.5 - 2 * BETA_MATLAB / (T_RUN * esc.N_REF ** 2 * fit[0]):.4f}.')
ellipse_table.round(4)
```

**The losing plants against the predicted ellipse.** Red: plants on which ESC loses (all of them). Black: the
predicted boundary, margin = $S_0$. Dotted: plants whose ideal ratio is 0.5 on average over the run. The steady-start
sets use $S_0 = L_\mathrm{dith}$ and no fitted number; the MATLAB start uses $S_0$ and $\beta$ measured on the grid below.

```python
def ellipse_panel(ax, name, sample=20000):
    r, S0, beta = cases[name]
    lost, won = r[r['gain'] < 0], r[r['gain'] >= 0]
    centre_u = -2 * beta / (T_RUN * esc.N_REF ** 2)
    half_u = 2.2 * np.sqrt(3) * np.sqrt(S0 / (T_RUN * esc.N_REF ** 2))
    half_l = 2.2 * np.sqrt(12 * S0 / (T_RUN * esc.N_REF ** 2)) / (0.5 * T_RUN)
    x_lim = (0.5 + centre_u - half_u, 0.5 + centre_u + half_u)
    y_lim = (1e3 * (-1.5 * centre_u / (0.5 * T_RUN) - half_l), 1e3 * (-1.5 * centre_u / (0.5 * T_RUN) + half_l))
    box = won[won['r_ideal_0'].between(*x_lim) & (1e3 * won['lambda_r_ideal']).between(*y_lim)]
    box = box.sample(min(sample, len(box)), random_state=0)
    ax.scatter(box['r_ideal_0'], 1e3 * box['lambda_r_ideal'], s=4, linewidths=0, color=esc.COLORS['light'], rasterized=True,
               label='ESC wins')
    ax.scatter(lost['r_ideal_0'], 1e3 * lost['lambda_r_ideal'], s=6, linewidths=0, color=esc.COLORS['bad'], rasterized=True,
               label='ESC loses')
    R, L = np.meshgrid(np.linspace(*x_lim, 400), np.linspace(*y_lim, 400) * 1e-3)
    ax.contour(R, 1e3 * L, predicted_margin(R, L, beta=beta) - S0, levels=[0], colors='black', linewidths=1.3)
    lam_line = np.linspace(*y_lim, 200) * 1e-3
    ax.plot(0.5 / mean_exp(lam_line, T_RUN), 1e3 * lam_line, color=esc.COLORS['grey'], ls=':', lw=1.2)
    ax.plot([], [], color='black', lw=1.3, label='predicted: margin = $S_0$')
    ax.plot([], [], color=esc.COLORS['grey'], ls=':', lw=1.2, label='run average of $r_\\mathrm{ideal}$ = 0.5')
    ax.set_xlim(*x_lim); ax.set_ylim(*y_lim)
    ax.set_xlabel(esc.label('r_ideal_0'))
    ax.set_title(name, fontsize=10)


# dpi=200: sharp when the notebook shrinks the figure
fig, axes = plt.subplots(2, 3, figsize=(16, 9), layout='constrained', dpi=200)
for ax, name in zip(axes.flat, cases):
    ellipse_panel(ax, name)
for row in axes:
    row[0].set_ylabel(r'drift rate $\lambda$ [10$^{-3}$ s$^{-1}$]')
handles, labels = axes[0, 0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 0.0), ncol=4, markerscale=3)
fig.suptitle('Plants on which ESC loses, and the predicted ellipse')
plt.show()
```

**Does the run length change the angle?** `OTO_ESC_losing_region.py` runs the reference set ($k$ = 1, $a$ = 0.2,

Prediction: with the steady start $S_0 = L_\mathrm{dith} \propto T$, so $\delta_*$ and the width in $r_{\mathrm{ideal},0}$ stay
the same, while the extent in $\lambda$ and the slope of the long axis scale with $1/T$.

```python
GRID_CASES = {'steady start, T = 50 s': 'steady-T50', 'steady start, T = 100 s': 'steady-T100',
              'steady start, T = 200 s': 'steady-T200', 'MATLAB start, T = 100 s': 'matlab-T100'}
grids, grid_rows = {}, {}
for label, case in GRID_CASES.items():
    g = pd.read_csv(f'OTO-losing_region-{case}.csv')
    g['gain'] = g['energy_ESC'] - g['energy_const']
    T_case = float(g['tend'].iloc[0])
    at_centre = np.isclose(g['r_ideal_0'], 0.5, rtol=0, atol=1e-9) & np.isclose(g['lambda_r_ideal'], 0, rtol=0, atol=1e-12)
    S0 = float((g['energy_ideal'] - g['energy_ESC'])[at_centre].iloc[0])
    beta = BETA_MATLAB if case.startswith('matlab') else 0.0
    grids[label] = (g, T_case, S0, beta)
    lost = g[g['gain'] < 0]
    on_axis = lost[np.isclose(lost['lambda_r_ideal'], 0, rtol=0, atol=1e-12)]
    slope, ratio = losing_axes(lost['r_ideal_0'], lost['lambda_r_ideal'], T_case)
    d_star = np.sqrt(S0 / (T_case * esc.N_REF ** 2))
    grid_rows[label] = {'S0': S0, 'dither loss L_dith': float(esc.dither_loss(0.2, 6, T=T_case)),
                        'delta_* (predicted half-width)': d_star,
                        'half-width at lambda = 0': (on_axis['r_ideal_0'].max() - on_axis['r_ideal_0'].min()) / 2,
                        'predicted lambda extent [1e-3/s]': 1e3 * np.sqrt(12) * d_star / (0.5 * T_case),
                        'lambda range [1e-3/s]': f"{1e3 * lost['lambda_r_ideal'].min():.2f} to {1e3 * lost['lambda_r_ideal'].max():.2f}",
                        'slope in plot, predicted [1e-3/s per unit]': 1e3 * PRED_SLOPE / (0.5 * T_case),
                        'slope in plot, measured [1e-3/s per unit]': 1e3 * slope / (0.5 * T_case),
                        'axis ratio': ratio,
                        'centre r_ideal_0': lost['r_ideal_0'].mean(), 'centre lambda [1e-3/s]': 1e3 * lost['lambda_r_ideal'].mean()}
print('Grid step: 0.0005 in r_ideal_0, 0.02e-3 1/s in lambda.')
pd.DataFrame(grid_rows).T.round(4)
```

**Reading the results**

- **Same shape everywhere.** For all five chosen sets, for the MATLAB start and for all three run lengths the losing
  plants form an ellipse with a long-axis slope of about $-1.8$ to $-2.2$ in $(u, v)$ and an axis ratio of about 3.6 to
  5.3 (prediction: $-1.87$ and 4.4; the sets with only 12 to 28 losing plants give the noisiest numbers). The rule
  "ESC loses where the predicted margin is below $S_0$" classifies 99.82 % (MATLAB start) and 99.97 to 100.00 %
  (steady start) of the 100,000 random plants correctly. The predicted ellipse is a few percent too small, because
  the margin formula overestimates the true margin near 0.5 by 2 to 6 %. With the MATLAB start the true region also
  reaches a little further at its upper-left end, where the start-up costs ESC slightly more.
- **The angle comes from the run length.** With the steady start the measured slope in the plot is $-79$, $-39$ and
  $-19.5 \cdot 10^{-3}$ s$^{-1}$ per unit for $T$ = 50, 100 and 200 s (predicted $-75$, $-37$, $-18.7$), and the extent
  in $\lambda$ halves when $T$ doubles. The width in $r_{\mathrm{ideal},0}$ stays at $\pm 0.002$ (predicted
  $\delta_* = 0.0023$), because the dither loss grows with $T$ just like the margin.
- **The size comes from $S_0$.** The ESC parameters change only the size: $\delta_*$ = 0.0004 (`fastest`) to 0.0022
  (reference) for the steady start. The MATLAB start adds a start-up cost of about 0.2 energy units to the dither loss
  (0.066), so $S_0$ = 0.27 and the ellipse is twice as wide.
- **The position comes from the start.** With the MATLAB start the centre moves to $r_{\mathrm{ideal},0} \approx 0.494$,
  $\lambda \approx +0.17 \cdot 10^{-3}$ s$^{-1}$, along the long axis, as predicted from the start asymmetry $\beta$
  (predicted centre 0.4947). With the steady start the ellipse is centred on (0.5, 0).
- A circle would need the starting offset and the drift to affect the margin independently. They do not: the fixed ratio
  only cares how far the ideal ratio is from 0.5 during the run, and a drift towards 0.5 makes up for a starting offset.
